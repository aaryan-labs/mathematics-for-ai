## 1. Why this matters

Feature sets can contain redundant variables, and learned representations occupy subspaces. Span, independence and basis explain what can be represented and how many independent directions exist.

## 2. Concepts and mathematical foundations

A **linear combination** of vectors $v_1,\ldots,v_k$ is $c_1v_1+\cdots+c_kv_k$. The **span** is the set of all such combinations.

A vector space is closed under vector addition and scalar multiplication and contains the zero vector and additive inverses. A subspace is a subset that is itself a vector space.

Vectors are **linearly independent** if $c_1v_1+\cdots+c_kv_k=0$ implies every coefficient is zero. Otherwise they are dependent. A **basis** is a linearly independent set that spans the space. The number of vectors in a basis is the **dimension**.

## 3. Derivation and worked example

Let $v_1=(1,0)$ and $v_2=(0,1)$. Then

$$3v_1-2v_2=(3,-2)$$

Every vector $(x,y)$ can be written as $xv_1+yv_2$, so these vectors span $\mathbb{R}^2$. They are independent because $c_1(1,0)+c_2(0,1)=(0,0)$ forces $c_1=c_2=0$. Therefore they form a basis and $\dim(\mathbb{R}^2)=2$.

## Python / NumPy verification

Work through the code after attempting the calculation manually.

```python
import numpy as np

v1 = np.array([1.0, 0.0])
v2 = np.array([0.0, 1.0])
combination = 3 * v1 - 2 * v2
B = np.column_stack([v1, v2])

print("linear combination:", combination)
print("basis matrix shape:", B.shape)
print("rank:", np.linalg.matrix_rank(B))

# A dependent set: the third vector is the sum of the first two.
C = np.column_stack([v1, v2, v1 + v2])
print("dependent-set rank:", np.linalg.matrix_rank(C), "number of vectors:", C.shape[1])
```

## AI/ML connection

Independent features add distinct directions of information; dependent features introduce redundancy. Embedding dimensions define a representation space, while basis concepts explain alternative coordinate systems.

## Assumptions, limitations and common mistakes

Numerical independence tests use tolerances. Near-dependence can be harder to detect than exact dependence. A set can span a space without being a basis if it is redundant.
