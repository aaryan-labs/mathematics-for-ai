## 1. Why this matters

An ML dataset, an embedding table and a neural-network batch are all represented numerically. Before operating on them, we need to know what each axis means and how dimensions fit together.

## 2. Concepts and mathematical foundations

* A **scalar** is one number.
* A **vector** is an ordered collection of numbers, often written as $\mathbf{x}\in\mathbb{R}^n$.
* A **matrix** is a two-dimensional array, $A\in\mathbb{R}^{m\times n}$.
* A **tensor** is a general multi-dimensional array.

The **shape** records the size along each axis. For a dataset matrix $X\in\mathbb{R}^{m\times n}$, a common convention is $m$ observations and $n$ features. Shape is not the same as mathematical rank: tensor order counts axes; matrix rank counts independent rows or columns.

A vector may be represented as a column vector ($n\times1$) or row vector ($1\times n$). Keep the convention explicit because matrix multiplication depends on shape.

## 3. Derivation and worked example

Suppose three customers are represented by two features: transaction count and average transaction value.

$$X=\begin{bmatrix}4&120\\7&85\\2&210\end{bmatrix}\in\mathbb{R}^{3\times2}$$

There are 3 observations and 2 features. The second observation is $[7,85]$. A single observation can be represented as a vector in $\mathbb{R}^2$.

## Python / NumPy verification

Work through the code after attempting the calculation manually.

```python
import numpy as np

x = np.array([4, 120])                  # one observation, shape (2,)
X = np.array([[4, 120], [7, 85], [2, 210]])  # dataset, shape (3, 2)
batch = np.zeros((2, 3, 4))             # example rank-3 tensor

for name, value in {"x": x, "X": X, "batch": batch}.items():
    print(name, "shape =", value.shape, "ndim =", value.ndim)

print("Second observation:", X[1])
print("Feature column 0:", X[:, 0])
```

## AI/ML connection

Feature vectors describe individual examples; dataset matrices collect examples; embedding matrices store many learned representations; batches of images or token sequences naturally require higher-order tensors.

## Assumptions, limitations and common mistakes

Do not confuse the number of axes with matrix rank. Do not assume every dataset uses rows as observations without checking its convention. A one-dimensional NumPy array has shape `(n,)`, not `(n, 1)`.
