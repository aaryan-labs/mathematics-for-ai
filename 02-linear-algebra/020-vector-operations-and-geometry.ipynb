## 2. Concepts and mathematical foundations

For vectors $\mathbf{a},\mathbf{b}\in\mathbb{R}^n$, addition and scalar multiplication are componentwise. The Euclidean norm is $\|\mathbf{x}\|_2=\sqrt{\sum_i x_i^2}$; the Manhattan norm is $\|\mathbf{x}\|_1=\sum_i|x_i|$.

The dot product is $\mathbf{a}\cdot\mathbf{b}=\sum_i a_ib_i$. Geometrically,

$$\mathbf{a}\cdot\mathbf{b}=\|\mathbf{a}\|\|\mathbf{b}\|\cos\theta$$

Thus a positive dot product indicates an acute angle, zero indicates orthogonality (for non-zero vectors, 90°), and a negative dot product indicates an obtuse angle. Cosine similarity divides the dot product by both norms, focusing on direction rather than magnitude.

## 3. Derivation and worked example

Let $\mathbf{a}=(2,1)$ and $\mathbf{b}=(1,3)$.

$$\mathbf{a}\cdot\mathbf{b}=2(1)+1(3)=5$$

$$\|\mathbf{a}\|_2=\sqrt5,\quad \|\mathbf{b}\|_2=\sqrt{10}$$

$$\cos\theta=\frac{5}{\sqrt5\sqrt{10}}=\frac{1}{\sqrt2}$$

So $\theta=45^\circ$. The Euclidean distance is $\sqrt{(2-1)^2+(1-3)^2}=\sqrt5$.

## Distance and similarity

Euclidean distance: $d_2(a,b)=\|a-b\|_2$. Manhattan distance: $d_1(a,b)=\|a-b\|_1$. Cosine similarity: $\frac{a\cdot b}{\|a\|_2\|b\|_2}$. Cosine distance is often defined as one minus cosine similarity, but conventions vary. Cosine similarity is undefined for a zero vector.

## Python / NumPy verification

Work through the code after attempting the calculation manually.

```python
import numpy as np

a = np.array([2.0, 1.0])
b = np.array([1.0, 3.0])

dot = np.dot(a, b)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)
cos_sim = dot / (norm_a * norm_b)
angle_deg = np.degrees(np.arccos(np.clip(cos_sim, -1.0, 1.0)))

print("dot product:", dot)
print("L2 norms:", norm_a, norm_b)
print("cosine similarity:", cos_sim)
print("angle in degrees:", angle_deg)
print("Euclidean distance:", np.linalg.norm(a - b))
print("Manhattan distance:", np.linalg.norm(a - b, ord=1))
```

## AI/ML connection

Vector norms support normalisation and regularisation. Dot products appear in linear prediction and similarity scoring. Cosine similarity is widely used to compare embeddings in semantic retrieval; Euclidean distance is used in nearest-neighbour methods.

## Assumptions, limitations and common mistakes

Cosine similarity ignores magnitude. Euclidean distance is scale-sensitive. Feature scaling can change distance-based results. Floating-point cosine values may drift slightly outside [-1, 1], so clip before applying arccos. Zero vectors need special handling.
