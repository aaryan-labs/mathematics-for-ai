## 1. Why this matters

AI engineering frequently manipulates high-dimensional representations. Correctly reasoning about shapes, transformations and similarity is essential for implementing models and diagnosing dimension errors.

## 2. Concepts and mathematical foundations

A dataset is often represented by $X\in\mathbb{R}^{N\times d}$: $N$ observations and $d$ features. An embedding is a vector $e\in\mathbb{R}^d$; a collection of embeddings forms a matrix.

A batch of row-vector inputs $X\in\mathbb{R}^{B\times d_{in}}$ transformed by weights $W\in\mathbb{R}^{d_{in}\times d_{out}}$ gives

$$Z=XW+b$$

where $b\in\mathbb{R}^{d_{out}}$ is broadcast across the batch.

For Transformer-style projections, if token representations are $X\in\mathbb{R}^{T\times d_{model}}$ and projection matrices have shapes $W_Q,W_K,W_V\in\mathbb{R}^{d_{model}\times d_k}$ (with value output dimension possibly $d_v$), then $Q=XW_Q$, $K=XW_K$, $V=XW_V$. The product $QK^T$ has shape $T\times T$. This notebook covers the linear algebra and dimensions, not softmax or the complete attention mechanism.

## 3. Derivation and worked example

Suppose a batch contains $B=4$ examples, each with $d_{in}=3$ features, and the layer produces $d_{out}=2$ values. Then $X$ has shape $(4,3)$, $W$ has shape $(3,2)$, and $XW$ has shape $(4,2)$. The bias has shape $(2,)$.

For a sequence of $T=5$ tokens with model dimension 8 and projection dimension 4: $X$ is $(5,8)$, each projection matrix is $(8,4)$, and $Q,K,V$ are each $(5,4)$. Thus $QK^T$ is $(5,5)$.

## Embedding similarity

Given query embedding $q$ and document embeddings stored as rows of $E$, dot-product scores are $Eq$. For cosine similarity, normalise vectors first or divide each dot product by the relevant norms. Rank scores to retrieve the closest candidates. Similarity is a mathematical score, not proof of semantic correctness.

## Numerical computing

Floating-point arithmetic is approximate. Use `np.allclose` for numerical equality, inspect residuals, consider conditioning, and prefer vectorised operations over Python loops when appropriate. Broadcasting follows shape rules and should be checked explicitly.

## Python / NumPy verification

Work through the code after attempting the calculation manually.

```python
import numpy as np

# Batch of feature vectors through an affine layer
rng = np.random.default_rng(7)
X = rng.normal(size=(4, 3))
W = rng.normal(size=(3, 2))
bias = rng.normal(size=2)
Z = X @ W + bias
print("linear layer output shape:", Z.shape)

# Embedding similarity: rows of E are document embeddings
E = rng.normal(size=(6, 4))
q = rng.normal(size=4)
E_norm = E / np.linalg.norm(E, axis=1, keepdims=True)
q_norm = q / np.linalg.norm(q)
cosine_scores = E_norm @ q_norm
ranking = np.argsort(cosine_scores)[::-1]
print("cosine scores:", cosine_scores)
print("ranked document indices:", ranking)

# Transformer projection dimensions
T, d_model, d_proj = 5, 8, 4
X_tokens = rng.normal(size=(T, d_model))
Wq = rng.normal(size=(d_model, d_proj))
Wk = rng.normal(size=(d_model, d_proj))
Wv = rng.normal(size=(d_model, d_proj))
Q, K, V = X_tokens @ Wq, X_tokens @ Wk, X_tokens @ Wv
scores = Q @ K.T
print("Q, K, V shapes:", Q.shape, K.shape, V.shape)
print("QK.T shape:", scores.shape)
```
