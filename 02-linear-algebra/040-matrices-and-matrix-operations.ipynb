## 2. Concepts and mathematical foundations

A matrix $A\in\mathbb{R}^{m\times n}$ has $m$ rows and $n$ columns. Addition requires equal shapes. If $A$ is $m\times n$ and $B$ is $n\times p$, then $AB$ is $m\times p$.

Matrix multiplication is not generally commutative: $AB\ne BA$. The transpose swaps rows and columns; $(AB)^T=B^TA^T$. Identity matrices satisfy $AI=IA=A$ where dimensions permit.

For a $2\times2$ matrix $A=\begin{bmatrix}a&b\\c&d\end{bmatrix}$, $\det(A)=ad-bc$. A square matrix is invertible exactly when its determinant is non-zero (equivalently, it has full rank).

## 3. Derivation and worked example

Let $A=\begin{bmatrix}1&2\\3&4\end{bmatrix}$ and $B=\begin{bmatrix}2&0\\1&2\end{bmatrix}$.

$$AB=\begin{bmatrix}4&4\\10&8\end{bmatrix}$$

$$\det(A)=1(4)-2(3)=-2$$

Since the determinant is non-zero, $A$ is invertible. Its inverse is $A^{-1}=\frac{1}{-2}\begin{bmatrix}4&-2\\-3&1\end{bmatrix}$.

## Python / NumPy verification

Work through the code after attempting the calculation manually.

```python
import numpy as np

A = np.array([[1.0, 2.0], [3.0, 4.0]])
B = np.array([[2.0, 0.0], [1.0, 2.0]])

print("A @ B =\n", A @ B)
print("A transpose =\n", A.T)
print("det(A) =", np.linalg.det(A))
print("inverse(A) =\n", np.linalg.inv(A))
print("check A @ inverse(A) =\n", A @ np.linalg.inv(A))
```

## AI/ML connection

Weight matrices transform feature vectors. Transposes appear in gradient expressions and covariance calculations. Matrix multiplication efficiently processes batches and composes linear maps.

## Assumptions, limitations and common mistakes

Only square matrices have ordinary two-sided inverses. A determinant is defined for square matrices. Avoid using an explicit inverse to solve numerical systems; use a solver. Matrix multiplication and elementwise multiplication are different operations.
