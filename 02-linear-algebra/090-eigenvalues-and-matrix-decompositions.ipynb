## 1. Why this matters

Eigenvectors identify directions preserved by a transformation up to scale. SVD exposes dominant directions and supports low-rank approximation, compression and many numerical algorithms.

## 2. Concepts and mathematical foundations

An eigenvector $v\ne0$ of a square matrix $A$ satisfies $Av=\lambda v$, where $\lambda$ is its eigenvalue. Eigenvalues solve $\det(A-\lambda I)=0$.

If a matrix has enough linearly independent eigenvectors, it can be diagonalised: $A=PDP^{-1}$. Real symmetric matrices have real eigenvalues and an orthonormal eigenbasis.

The singular value decomposition is

$$A=U\Sigma V^T$$

where columns of $U$ and $V$ are orthonormal and $\Sigma$ contains non-negative singular values. QR decomposition writes $A=QR$, with orthonormal columns in $Q$ and upper-triangular $R$ (for the standard full-column-rank case).

## 3. Derivation and worked example

For $A=\begin{bmatrix}2&0\\0&3\end{bmatrix}$, the coordinate axes are eigenvector directions. $A(1,0)=2(1,0)$ and $A(0,1)=3(0,1)$, so the eigenvalues are 2 and 3.

For SVD, singular values describe how strongly the matrix stretches orthogonal input directions. Keeping only the largest singular values gives a lower-rank approximation.

## Python / NumPy verification

Work through the code after attempting the calculation manually.

```python
import numpy as np

A = np.array([[2.0, 0.0], [0.0, 3.0]])
values, vectors = np.linalg.eig(A)
print("eigenvalues:", values)
print("eigenvector check A @ V = V @ diag(lambda):",
      np.allclose(A @ vectors, vectors @ np.diag(values)))

M = np.array([[3.0, 1.0], [1.0, 2.0], [0.0, 1.0]])
U, singular_values, Vt = np.linalg.svd(M, full_matrices=False)
Sigma = np.diag(singular_values)
reconstructed = U @ Sigma @ Vt
print("singular values:", singular_values)
print("SVD reconstruction error:", np.linalg.norm(M - reconstructed))

Q, R = np.linalg.qr(M)
print("QR reconstruction error:", np.linalg.norm(M - Q @ R))
print("Q.T @ Q =\n", Q.T @ Q)
```

## AI/ML connection

Eigenvectors and eigenvalues underpin PCA for symmetric covariance matrices. SVD supports dimensionality reduction, compression, denoising and stable least-squares solutions. QR is used in numerical linear algebra.

## Assumptions, limitations and common mistakes

Not every matrix is diagonalizable over the reals. Eigenvectors can be numerically sensitive, especially for non-normal matrices or repeated eigenvalues. Singular values are not the same as eigenvalues. Decomposition outputs may differ in sign while representing the same decomposition.
