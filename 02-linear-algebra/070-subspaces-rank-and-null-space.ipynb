## 1. Why this matters

A transformation may preserve some directions and collapse others. Rank and null space describe the output dimensions and the inputs that become zero.

## 2. Concepts and mathematical foundations

The **column space** is the span of the columns of $A$; it contains all possible outputs $Ax$. The **row space** is the span of its rows. The **null space** is the set of vectors satisfying $Ax=0$.

The **rank** is the dimension of the column space (equal to the dimension of the row space). The **nullity** is the dimension of the null space. For $A\in\mathbb{R}^{m\times n}$,

$$\operatorname{rank}(A)+\operatorname{nullity}(A)=n$$

This is the rank-nullity theorem.

## 3. Derivation and worked example

Consider $A=\begin{bmatrix}1&2\\2&4\end{bmatrix}$. Its second column is twice the first, so rank is 1. The equation $Ax=0$ reduces to $x_1+2x_2=0$, giving $x=t(-2,1)$. The null space has dimension 1. Thus rank + nullity = 1 + 1 = 2, the number of columns.

## Python / NumPy verification

Work through the code after attempting the calculation manually.

```python
import numpy as np

A = np.array([[1.0, 2.0], [2.0, 4.0]])
rank = np.linalg.matrix_rank(A)
u, s, vh = np.linalg.svd(A)
tol = max(A.shape) * np.finfo(float).eps * s[0]
nullity = A.shape[1] - np.sum(s > tol)

print("rank:", rank)
print("estimated nullity:", nullity)
print("rank + nullity:", rank + nullity)
print("singular values:", s)
print("candidate null-space direction:", vh[-1])
```

## AI/ML connection

Rank reveals redundant features and the effective dimension of a linear transformation. Null-space directions represent changes to an input that do not change the output, relevant to identifiability and information loss.

## Assumptions, limitations and common mistakes

Numerical rank depends on a tolerance, especially for nearly dependent columns. SVD-based null-space estimates are numerical approximations. Rank does not by itself explain the semantic meaning of a feature.
