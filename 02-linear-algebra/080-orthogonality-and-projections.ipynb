## 1. Why this matters

When a system has no exact solution, we often seek the closest achievable output. Projection and orthogonality explain why least-squares residuals are perpendicular to the model's column space.

## 2. Concepts and mathematical foundations

Vectors $u$ and $v$ are orthogonal when $u\cdot v=0$. An orthonormal set is mutually orthogonal with unit norms. A matrix with orthonormal columns satisfies $Q^TQ=I$.

The projection of $a$ onto non-zero $b$ is

$$\operatorname{proj}_b(a)=\frac{a\cdot b}{b\cdot b}b$$

For a matrix $Q$ with orthonormal columns, projection onto its column space is $QQ^Ty$. In least squares, choose $x$ to minimise $\|Ax-b\|_2^2$; at the solution, the residual is orthogonal to the column space of $A$.

## 3. Derivation and worked example

Project $a=(3,2)$ onto $b=(1,0)$:

$$\operatorname{proj}_b(a)=\frac{3}{1}(1,0)=(3,0)$$

The residual is $a-\operatorname{proj}_b(a)=(0,2)$, which is perpendicular to $(1,0)$ because their dot product is zero.

## Python / NumPy verification

Work through the code after attempting the calculation manually.

```python
import numpy as np

a = np.array([3.0, 2.0])
b = np.array([1.0, 0.0])
projection = (np.dot(a, b) / np.dot(b, b)) * b
residual = a - projection
print("projection:", projection)
print("residual:", residual)
print("residual dot direction:", np.dot(residual, b))

# Least-squares example
A = np.array([[1.0], [1.0], [1.0]])
y = np.array([1.0, 2.0, 4.0])
x_hat, *_ = np.linalg.lstsq(A, y, rcond=None)
r = y - A @ x_hat
print("least-squares estimate:", x_hat)
print("residual:", r)
print("A.T @ residual (approximately zero):", A.T @ r)
```

## AI/ML connection

Least squares is foundational to linear regression and parameter estimation. Orthogonal projections also appear in dimensionality reduction and numerical algorithms.

## Assumptions, limitations and common mistakes

The vector projection formula requires a non-zero direction. The simple $QQ^T$ formula assumes orthonormal columns. Normal equations can be numerically less stable than QR or SVD-based least-squares solvers.
