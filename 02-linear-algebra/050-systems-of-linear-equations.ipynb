## 1. Why this matters

Many estimation and modelling problems ask for unknown parameters that satisfy simultaneous linear relationships. Matrix form makes the structure explicit and enables reliable numerical solution.

## 2. Concepts and mathematical foundations

A system of linear equations can be written $A\mathbf{x}=\mathbf{b}$, where $A$ contains coefficients, $\mathbf{x}$ contains unknowns and $\mathbf{b}$ contains constants.

Gaussian elimination uses elementary row operations: swap rows, multiply a row by a non-zero scalar, or add a multiple of one row to another. These operations preserve the solution set. Echelon form enables back substitution.

A system may have one solution, infinitely many solutions or no solution. For numerical work, use a solver rather than explicitly computing $A^{-1}b$.

## 3. Derivation and worked example

Solve $x+y=5$ and $2x-y=1$.

Add the equations: $3x=6$, so $x=2$. Substitute into $x+y=5$: $y=3$.

Matrix form:

$$\begin{bmatrix}1&1\\2&-1\end{bmatrix}\begin{bmatrix}x\\y\end{bmatrix}=\begin{bmatrix}5\\1\end{bmatrix}$$

## Python / NumPy verification

Work through the code after attempting the calculation manually.

```python
import numpy as np

A = np.array([[1.0, 1.0], [2.0, -1.0]])
b = np.array([5.0, 1.0])

solution = np.linalg.solve(A, b)
print("solution:", solution)
print("reconstructed right-hand side:", A @ solution)
print("residual:", A @ solution - b)
print("condition number:", np.linalg.cond(A))
```

## AI/ML connection

Linear systems underpin parameter estimation, transformations and the least-squares foundations used in regression. Solvers also appear inside numerical optimisation and scientific computing.

## Assumptions, limitations and common mistakes

A direct solve requires a compatible, nonsingular square system for a unique solution. Real data may be noisy or overdetermined. Ill-conditioned systems can amplify small input errors. Floating-point calculations are approximate.
