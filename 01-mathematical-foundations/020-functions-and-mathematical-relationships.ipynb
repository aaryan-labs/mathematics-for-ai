## 1. What is a function?

A function assigns exactly one output to each permitted input. We write $y=f(x)$. The input is $x$; the output is $f(x)$.

Example: $f(x)=2x+3$. If $x=4$, then $f(4)=11$.

A function is a rule/mapping, not merely an equation or a plotted curve. Different inputs may map to the same output, but one input cannot map to two different outputs in an ordinary function.

```python
def f(x):
    return 2*x + 3

for value in [0, 1, 4]:
    print(f"f({value}) = {f(value)}")
```

## 2. Domain and range

The **domain** is the set of permitted inputs. The **range** is the set of outputs actually produced.

Examples:
- $f(x)=1/x$: real domain excludes zero.
- $g(x)=\sqrt{x}$: real domain requires $x\ge0$.
- $h(x)=x^2$, with real inputs: range is $[0,\infty)$.

Always consider the number system and stated restrictions. Domain and range are not interchangeable.

```python
import numpy as np
import matplotlib.pyplot as plt

x1 = np.linspace(-4, 4, 400)
x2 = np.linspace(-4, 4, 400)
x2 = x2[np.abs(x2) > 0.08]

plt.figure()
plt.plot(x1, x1**2, label="y = x^2")
plt.title("A quadratic function")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()

plt.figure()
plt.plot(x2, 1/x2, label="y = 1/x")
plt.title("A reciprocal function (zero excluded)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

## 3. Common function families

| Family | Example | Characteristic |
|---|---|---|
| Linear | $f(x)=mx+c$ | Constant slope |
| Quadratic | $f(x)=ax^2+bx+c$ | Parabolic graph |
| Polynomial | $f(x)=x^3-2x+1$ | Nonnegative integer powers |
| Rational | $f(x)=1/(x+1)$ | Ratio of polynomials; denominator restrictions |
| Exponential | $f(x)=a^x$ | Variable appears in exponent |
| Logarithmic | $f(x)=\log x$ | Inverse of an exponential (with matching base) |
| Piecewise | Rule changes by input region | Different formulae on different regions |

A function family describes mathematical behaviour; it does not by itself specify a model.

## 4. Graphs, slope and intercepts

For a straight line $y=mx+c$:
- $m$ is the slope: change in output per unit change in input.
- $c$ is the y-intercept: output at $x=0$.

For two points, slope is

$$
m=\frac{y_2-y_1}{x_2-x_1},\qquad x_2\ne x_1.
$$

A positive slope means the output increases as the input increases; a negative slope means it decreases. Slope is a foundational idea for derivatives, but derivatives are covered in the Calculus thread.

```python
# Plot a linear and quadratic function.
x = np.linspace(-5, 5, 300)
plt.figure()
plt.plot(x, 2*x + 1, label="2x + 1")
plt.plot(x, x**2 - 2, label="x^2 - 2")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.show()
```

## 5. Function transformations

Starting with $y=f(x)$:
- $f(x)+k$: shift vertically by $k$.
- $f(x-h)$: shift right by $h$.
- $af(x)$: vertical scaling; reflection if $a<0$.
- $f(bx)$: horizontal scaling; reflection if $b<0$.

The inside of the function can feel counterintuitive: $f(x-h)$ shifts the graph right, not left. Check by tracking a known point.

```python
x = np.linspace(-4, 6, 300)
plt.figure()
plt.plot(x, x**2, label="f(x) = x^2")
plt.plot(x, (x-2)**2 + 1, label="f(x-2) + 1")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

## 6. Composite functions

Composition means applying one function and then another:

$$
(g\circ f)(x)=g(f(x)).
$$

If $f(x)=2x+1$ and $g(x)=x^2$, then

$$
(g\circ f)(x)=(2x+1)^2.
$$

Order matters: generally $g(f(x))\ne f(g(x))$. This is a helpful way to think about a sequence of transformations, including layers in a neural network.

```python
def f(x):
    return 2*x + 1

def g(x):
    return x**2

print("g(f(3)) =", g(f(3)))
print("f(g(3)) =", f(g(3)))
```

## 7. Inverse functions

An inverse function reverses a mapping. If $f(a)=b$, then $f^{-1}(b)=a$.

For $f(x)=2x+3$:
$$
y=2x+3\Rightarrow x=\frac{y-3}{2}
\Rightarrow f^{-1}(x)=\frac{x-3}{2}.
$$

A function must be one-to-one on its domain to have an inverse function on its full range. For example, $x^2$ is not one-to-one over all real numbers; restricting its domain to $x\ge0$ makes the inverse $\sqrt{x}$.

```python
import sympy as sp
x = sp.symbols('x')
f_expr = 2*x + 3
inverse_expr = sp.solve(sp.Eq(sp.Symbol('y'), f_expr), x)[0]
print("Inverse expression in y:", inverse_expr)
print("Check composition:", sp.simplify((2*((x-3)/2)+3)))
```

## Where this appears in AI/ML

- A model can be viewed as a function mapping features to predictions.
- A neural network composes parameterised functions/layers.
- Activation functions such as sigmoid, tanh and ReLU transform intermediate values.
- Domain restrictions matter: for example, $\log(x)$ requires positive input in the real-valued setting.
- Function graphs help interpret sensitivity, saturation and output ranges.
