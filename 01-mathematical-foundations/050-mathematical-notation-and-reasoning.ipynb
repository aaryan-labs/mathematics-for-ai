## 1. Mathematical notation as a working language

ML papers and documentation compress long explanations into notation. You should be able to identify:
- variables and parameters;
- indices and dimensions;
- subscripts and superscripts;
- operators and conditions;
- the objects being summed or multiplied.

Examples: $x_i$ means the $i$-th element; $w_j$ may denote the $j$-th parameter; $x^{(i)}$ is often used for the $i$-th example (convention varies). Always inspect the author's definitions.

The meaning of a symbol can vary by context. Definitions take precedence over memorised interpretations.

## 3. Summation notation

$$
\sum_{i=1}^{n}x_i=x_1+x_2+\cdots+x_n.
$$

The index $i$ takes integer values from 1 through $n$. Example:

$$
\sum_{i=1}^{4}i^2=1^2+2^2+3^2+4^2=30.
$$

A common dataset expression is the sample mean:

$$
\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i.
$$

Here $n$ is the number of observations and $x_i$ is observation $i$.

```python
values = [2, 4, 6, 8]
total = sum(values)
mean = total / len(values)
print("Sum:", total)
print("Mean:", mean)
```

## 4. Product notation

$$
\prod_{i=1}^{n}a_i=a_1a_2\cdots a_n.
$$

For example, $\prod_{i=1}^{4}i=1\cdot2\cdot3\cdot4=24$.

Sums and products are different aggregation operations. In probability, a joint probability under independence may be expressed as a product; log transformation converts it into a sum.

```python
import math
values = [1, 2, 3, 4]
print("Product:", math.prod(values))
```

## 5. Indices, vectors of observations and parameters

Consider a dataset $x_1,x_2,\ldots,x_n$. The index identifies an observation. If each observation has $d$ features, one possible notation is $x_{ij}$, where:
- $i$ indexes the observation;
- $j$ indexes the feature.

For example, $x_{23}$ is the third feature of the second observation under this convention.

In ML papers, superscripts may denote layers, time steps, classes or examples. Never assume an index meaning without checking.

```python
# A small dataset: rows are observations, columns are features.
import numpy as np
X = np.array([
    [1.2, 5.0, 0.0],
    [0.7, 3.5, 1.0],
    [2.1, 4.2, 0.0]
])
print("x_23 (row 2, feature 3, 1-based) =", X[1, 2])
print("Number of observations =", X.shape[0])
print("Number of features =", X.shape[1])
```

## 6. Equality, equivalence and implication

- An equation such as $x+2=5$ is true only for particular values of $x$.
- An identity such as $(x+1)^2=x^2+2x+1$ is true for every real $x$.
- An implication $P\Rightarrow Q$ means that whenever $P$ is true, $Q$ follows.
- Equivalence $P\iff Q$ means both implications hold.

When solving equations, transformations should preserve the solution set. Dividing by an expression that could be zero may discard or introduce cases; state restrictions and check solutions.

## 7. Approximation, precision and numerical values

Exact values and floating-point approximations are not the same. For example, $1/10$ has a finite decimal representation mathematically, but many decimal fractions cannot be represented exactly in binary floating point.

Use tolerances when comparing floating-point computations rather than relying on exact equality.

```python
a = 0.1 + 0.2
print(a)
print(a == 0.3)
print(abs(a - 0.3) < 1e-12)
```

## 8. Reading a compact ML-style expression

Consider:

$$
J(\theta)=\frac{1}{n}\sum_{i=1}^{n}(f_\theta(x_i)-y_i)^2.
$$

Interpretation:
- $J$: objective/loss as a function of parameters;
- $\theta$: model parameters;
- $n$: number of observations;
- $i$: observation index;
- $x_i$: input for observation $i$;
- $y_i$: corresponding target;
- $f_\theta(x_i)$: model prediction;
- squared difference: squared prediction error;
- summation and division by $n$: average across observations.

This is a mean squared error-style objective. The expression is included to practise reading notation; full loss-function mathematics belongs in the Classical ML thread.

```python
# Compute the illustrative mean squared error.
predictions = np.array([2.5, 1.0, 4.0])
targets = np.array([3.0, 1.5, 2.0])
J = np.mean((predictions - targets)**2)
print(J)
```
