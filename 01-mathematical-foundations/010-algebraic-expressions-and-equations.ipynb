## 1. Algebraic expressions

An **algebraic expression** combines numbers, variables and operations. A variable represents a value that may change; a coefficient multiplies a variable; a constant has no variable.

For example, in \(7x^2-4x+3\):
- Terms: \(7x^2\), \(-4x\), \(3\)
- Coefficients: \(7\), \(-4\)
- Constant: \(3\)
- Variable: \(x\)

Terms are **like terms** when they have identical variable parts, including identical powers. Thus \(3x^2\) and \(-5x^2\) are like terms, but \(3x^2\) and \(3x\) are not.

### Simplifying by collecting like terms

\[
5x^2+3x-2+4x^2-7x+5-2x^2
=(5+4-2)x^2+(3-7)x+(-2+5)
=7x^2-4x+3.
\]

Only coefficients of like terms are combined.

```python
# Verify simplification numerically at several x values.
def original(x):
    return 5*x**2 + 3*x - 2 + 4*x**2 - 7*x + 5 - 2*x**2

def simplified(x):
    return 7*x**2 - 4*x + 3

for x in [-2, 0, 1, 3]:
    print(x, original(x), simplified(x))
    assert original(x) == simplified(x)
```

## 2. Operations and the distributive property

The distributive property is

\[
a(b+c)=ab+ac, \qquad a(b-c)=ab-ac.
\]

Example:

\[
3(2x-5)+4(x+2)
=6x-15+4x+8
=10x-7.
\]

A frequent error is distributing to only one term inside the brackets. The multiplier applies to every term.

```python
# Expand and evaluate the two equivalent expressions.
def bracket_form(x):
    return 3*(2*x - 5) + 4*(x + 2)

def expanded_form(x):
    return 10*x - 7

for x in [-3, 0, 2.5, 10]:
    assert bracket_form(x) == expanded_form(x)
print("Equivalent for all tested values.")
```

## 3. Expansion and factorisation

Expansion removes brackets; factorisation expresses a sum/product as a product of simpler factors.

For two binomials:

\[
(x+a)(x+b)=x^2+(a+b)x+ab.
\]

To factorise \(x^2+7x+12\), find numbers whose:
- product is \(12\)
- sum is \(7\)

Those numbers are 3 and 4:

\[
x^2+7x+12=(x+3)(x+4).
\]

Factorisation is useful for identifying roots and simplifying expressions. Not every polynomial factors neatly over the integers.

```python
import sympy as sp
x = sp.symbols('x')
expr = x**2 + 7*x + 12
print("Expanded:", sp.expand((x + 3)*(x + 4)))
print("Factorised:", sp.factor(expr))
print("Roots:", sp.solve(expr, x))
```

## 4. Equations and rearrangement

An equation states that two expressions have equal value. To preserve equality, perform the same valid operation on both sides.

Example:

\[
8=\frac{5x-2}{4}
\Rightarrow 32=5x-2
\Rightarrow 34=5x
\Rightarrow x=6.8.
\]

A solution is a value that makes the original equation true. Always substitute it back when practical.

### Quadratic equations

A quadratic has form \(ax^2+bx+c=0\), with \(a\ne0\). Its roots can be found by factorisation or the quadratic formula:

\[
x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}.
\]

The discriminant \(b^2-4ac\) indicates the nature of real roots.

```python
# Solve and verify a linear and a quadratic equation.
x = sp.symbols('x')
print(sp.solve(sp.Eq(8, (5*x - 2)/4), x))
print(sp.solve(sp.Eq(x**2 + 7*x + 12, 0), x))
```

## 5. Inequalities

Inequalities compare values using \(<,\le,>,\ge\). Addition or subtraction of the same value preserves the direction. Multiplication or division by a **negative** value reverses it.

Example:

\[
-2x<6 \quad\Rightarrow\quad x>-3.
\]

This reversal is essential. Inequalities describe ranges of valid values rather than necessarily a single solution.

```python
# Numerical check of the solution to -2x < 6, namely x > -3.
for value in [-4, -3, -2, 0]:
    print(value, (-2*value < 6), (value > -3))
```

## 6. Fractions, ratios and proportions

A fraction \(a/b\) represents a quotient, with \(b\ne0\). Equivalent fractions result from multiplying numerator and denominator by the same nonzero value.

A ratio compares quantities. A proportion states that two ratios are equal:

\[
\frac{a}{b}=\frac{c}{d}
\quad\Rightarrow\quad ad=bc
\]

provided denominators are nonzero.

In ML, ratios appear in rates, proportions, normalisation and weighted combinations. Be careful to distinguish a ratio from a difference.

```python
# A simple proportion: 3/5 = x/20
x_value = 3 * 20 / 5
print(x_value)
assert 3/5 == x_value/20
```
