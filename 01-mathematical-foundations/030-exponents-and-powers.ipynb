## 1. Exponent fundamentals

In \(a^n\), \(a\) is the base and \(n\) is the exponent. For a positive integer \(n\), it means multiplying \(a\) by itself \(n\) times.

\[
a^4=a\cdot a\cdot a\cdot a.
\]

Exponents compactly represent repeated multiplication and are central to growth, scaling and numerical representation.

## 2. Laws of exponents

For valid nonzero bases and integer exponents:

\[
a^m a^n=a^{m+n},\qquad
\frac{a^m}{a^n}=a^{m-n},
\]
\[
(a^m)^n=a^{mn},\qquad
(ab)^n=a^n b^n,
\]
\[
a^0=1\ (a\ne0),\qquad
a^{-n}=\frac{1}{a^n}.
\]

The rules follow from counting repeated factors. For example, \(a^3a^2\) contains five factors of \(a\), so it is \(a^5\). Do not add exponents when adding terms: \(a^2+a^3\) cannot be simplified to \(a^5\).

```python
# Verify exponent laws numerically for a nonzero positive base.
a, m, n = 2, 3, 4
assert a**m * a**n == a**(m+n)
assert a**m / a**n == a**(m-n)
assert (a**m)**n == a**(m*n)
assert (a*3)**n == a**n * 3**n
print("Exponent identities verified for this example.")
```

## 3. Fractional exponents and roots

For positive real \(a\):

\[
a^{1/n}=\sqrt[n]{a},\qquad
a^{m/n}=\left(\sqrt[n]{a}\right)^m.
\]

Examples: \(16^{1/2}=4\), \(8^{1/3}=2\), and \(27^{2/3}=9\).

Be careful with even roots of negative real numbers: they are not real. More generally, exponent laws have domain restrictions depending on the base and exponent.

```python
values = [16**0.5, 8**(1/3), 27**(2/3), 2**-3]
print(values)
```

## 4. Exponential functions and growth

An exponential function has the variable in the exponent, commonly \(f(x)=a^x\) for \(a>0\), \(a\ne1\).

- If \(a>1\), it increases as \(x\) increases.
- If \(0<a<1\), it decreases.
- \(a^0=1\), so the graph passes through \((0,1)\).

Exponential growth compounds: multiplying by a fixed factor repeatedly creates a rapidly changing sequence.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 300)
plt.figure()
plt.plot(x, 2.0**x, label="2^x")
plt.plot(x, 0.5**x, label="(0.5)^x")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Exponential growth and decay")
plt.grid(True)
plt.legend()
plt.show()
```

## 5. Scientific notation and scale

Scientific notation writes a number as \(c\times10^n\), where typically \(1\le |c|<10\).

Examples:
- \(450000=4.5\times10^5\)
- \(0.00032=3.2\times10^{-4}\)

It helps communicate orders of magnitude and reason about very large or small quantities encountered in datasets and numerical computation.

```python
numbers = [450000, 0.00032, 6.02e23, 1.6e-19]
for number in numbers:
    print(f"{number:.3e}")
```
